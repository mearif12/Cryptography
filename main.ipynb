```python
# # #### E1-6
# # def matrix(key):
# #     s = ''.join(dict.fromkeys(
# #         (key + 'ABCDEFGHIKLMNOPQRSTUVWXYZ').upper().replace('J','I')))
# #     return [s[i:i+5] for i in range(0,25,5)]

# # def playfair(text, key, dec=False):
# #     text = text.upper().replace('J','I').replace(' ','')
# #     if not dec:
# #         p=[]; i=0
# #         while i < len(text):
# #             a=text[i]; b=text[i+1] if i+1<len(text) else 'X'
# #             p.append(a+'X' if a==b else a+b)
# #             i += 1 if a==b else 2
# #     else:
# #         p=[text[i:i+2] for i in range(0,len(text),2)]

# #     out=''
# #     for a,b in p:
# #         r1,c1=next((r,c) for r in range(5) for c in range(5) if m[r][c]==a)
# #         r2,c2=next((r,c) for r in range(5) for c in range(5) if m[r][c]==b)

# #         if r1==r2:
# #             d=-1 if dec else 1
# #             out += m[r1][(c1+d)%5]+m[r2][(c2+d)%5]
# #         elif c1==c2:
# #             d=-1 if dec else 1
# #             out += m[(r1+d)%5][c1]+m[(r2+d)%5][c2]
# #         else:
# #             out += m[r1][c2]+m[r2][c1]
# #     return out

# # key=input("Key: ")
# # msg=input("Plaintext: ")

# # m = matrix(key)
# # print("Matrix:")
# # for r in m:
# #     print(*r)

# # enc=playfair(msg,key)
# # print("Encrypted:",enc)
# # print("Decrypted:",playfair(enc,key,True))



# Short Theory — Playfair Cipher
# Playfair Cipher is a digraph substitution cipher that encrypts two letters at a time using a 5×5 matrix. In the matrix, J is usually combined with I.

# Important Points
# A 5×5 matrix is created using the given key.

# Duplicate letters are removed from the key.

# J is replaced by I.

# Plaintext is divided into pairs of letters.

# If two letters in a pair are the same, X is inserted between them.

# If one letter is left at the end, X is added.

# Encryption rules depend on whether the two letters are in the same row, same column, or different rows and columns.

# Important Rules
# 1. Same Row:
# Move each letter one position to the right.

# 2. Same Column:
# Move each letter one position downward.

# 3. Different Row and Column:
# Take the letters at the opposite corners of the rectangle.

# For decryption, the row and column movements are reversed.

# Matrix
# Given matrix:

# M O N A R
# C H Y B D
# E F G I K
# L P Q S T
# U V W X Z

# Mathematical/Pair Calculation
# Plaintext:

# INSTRUMENTS

# Divide into pairs:

# IN ST RU ME NT SX

# Here, X is added at the end to make a complete pair.

# For example, pair IN:

# I is at row 3, column 4.
# N is at row 1, column 3.

# They are in different rows and columns, so we use the rectangle rule.

# I → G
# N → A

# Therefore:

# IN → GA

# Applying the same Playfair rules to all pairs gives:

# IN ST RU ME NT SX

# →

# GA TL MZ CL RQ XA

# Therefore:

# Encrypted = GATLMZCLRQXA

# For decryption, the reverse rules are applied:

# GATLMZCLRQXA → INSTRUMENTSX

# The final X is padding added during encryption.

# Output
# Matrix:
# M O N A R
# C H Y B D
# E F G I K
# L P Q S T
# U V W X Z

# Encrypted: GATLMZCLRQXA
# Decrypted: INSTRUMENTSX
```
